```python
%load_ext autoreload
%autoreload 2

import caltool.calibrate as cal
import caltool.utils as calutil
from pypulsepal import PulsePalObject
```

```python
pulsepal = PulsePalObject('COM13')
```

```python
OUTPUT_CHANNEL = 1
TEST_VOLTAGES = [0, .1, .2, .3, .4, .5, .75, 1, 2, 3, 4, 5]

DEVICE_NAME = 'HM010_3'
```

```python
# run the calibration routine
measured_values = []
for V in TEST_VOLTAGES:
    pulsepal.setFixedVoltage(OUTPUT_CHANNEL,V) 
    print(V)
    #pulsepal.triggerOutputChannels(1, 1, 0, 0)
    measured_value = calutil.get_float(f'Enter the intensity in mW for voltage {V}')
    measured_values.append(measured_value)
print('Test voltages:', TEST_VOLTAGES)
print('Measured values:', measured_values)
```

```python
cal.create_calibration_curve(DEVICE_NAME, TEST_VOLTAGES, measured_values, polyorder=2, overwrite=False)
```

```python
cal.apply_calibration_curve(DEVICE_NAME, 2.5)
```

```python
pulsepal.setFixedVoltage(OUTPUT_CHANNEL,0) 
del pulsepal
```
